# Read DANRA data from an object store

```python
from pathlib import Path
import xarray as xr
import numpy as np
import os
```

```python
credentials_key = os.getenv("AWS_ACCESS_KEY_ID")
credentials_secret = os.getenv("AWS_SECRET_ACCESS_KEY")
```

```python
endpoint_url = os.getenv("ENDPOINT_URL")

ds_height_levels = xr.open_zarr( "s3://danra/v0.4.0/height_levels.zarr/",
   consolidated=True,
   storage_options={
       "key": credentials_key,
       "secret": credentials_secret,
       "client_kwargs": {"endpoint_url": endpoint_url},
    },
)

ds_height_levels
```

```python
ds_single_levels = xr.open_zarr( "s3://danra/v0.4.0/single_levels.zarr/",
   consolidated=True,
   storage_options={
       "key": credentials_key,
       "secret": credentials_secret,
       "client_kwargs": {"endpoint_url": endpoint_url},
    },
)

ds_single_levels
```
